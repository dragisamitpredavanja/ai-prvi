```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt
from google.colab import files

# 1. Učitavamo CIFAR-10 bazu podataka (slike u boji)
cifar10 = keras.datasets.cifar10
(x_trening, y_trening), (x_test, y_test) = cifar10.load_data()

# Nazivi klasa pošto su u bazi samo brojevi od 0 do 9
nazivi_klasa = ['avion', 'automobil', 'ptica', 'macka', 'jelen', 'pas', 'zaba', 'konj', 'brod', 'kamion']

# 2. Normalizacija piksela (sa 0-255 na opseg 0-1)
x_trening, x_test = x_trening / 255.0, x_test / 255.0

# 3. Pravimo Konvolucionu neuronsku mrežu (CNN) - najbolju arhitekturu za slike u boji
model = keras.Sequential([
    keras.layers.Input(shape=(32, 32, 3)), # 3 označava RGB kanale (boje)

    # Prvi konvolucioni sloj koji traži ivice i oblike na slikama
    keras.layers.Conv2D(32, (3, 3), activation='relu'),
    keras.layers.MaxPooling2D((2, 2)),

    # Drugi konvolucioni sloj za složenije teksture
    keras.layers.Conv2D(64, (3, 3), activation='relu'),
    keras.layers.MaxPooling2D((2, 2)),

    # Ispravljanje u 1D niz i slanje u klasične Dense slojeve
    keras.layers.Flatten(),
    keras.layers.Dense(64, activation='relu'),
    keras.layers.Dense(10, activation='softmax')
])

# 4. Kompajliranje modela
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# 5. Treniramo AI kroz 5 epoha
print("Treniranje na slikama u boji je počelo...")
model.fit(x_trening, y_trening, epochs=5, batch_size=64)
print("Treniranje završeno!")

# 6. Provera tačnosti na test podacima
gubitak, tacnost = model.evaluate(x_test, y_test, verbose=0)
print(f"\nTačnost na slikama u boji je: {tacnost * 100:.2f}%")

# 7. Testiramo AI na prvoj slici iz testa (indeks 0)
slika_za_test = x_test[0]
pravi_indeks = y_test[0][0]

predikcija = model.predict(slika_za_test[np.newaxis, ...], verbose=0)
pogodjeni_indeks = np.argmax(predikcija)

print(f"\n[AI PREDVIĐA]: Na slici je {nazivi_klasa[pogodjeni_indeks]}")
print(f"[TAČAN ODGOVOR]: Na slici je zapravo {nazivi_klasa[pravi_indeks]}")

# 8. Prikazujemo sliku na ekranu da proverimo
plt.figure(figsize=(2,2))
plt.imshow(slika_za_test)
plt.title(f"AI: {nazivi_klasa[pogodjeni_indeks]}")
plt.axis('off')
plt.show()
# Snimamo ceo model (arhitekturu i istrenirane težine) u jedan fajl
model.save('cifar10_model.h5')
print("Model je uspešno sačuvan!")
```
